# Control de flujo
## Condicionales y bucle for


```python

# Condicionales

if edad >= 18:
    print("Mayor de edad")
elif edad >= 13:
    print("Adolescente")
else:
    print("Niño")


# Bucles 

# for — cuando sabes cuántas veces repetir (o recorres algo) Se usa para recorrer una secuencia (lista, string, rango de números, etc.) elemento por elemento.


for i in range(5):
    print(i)
    # output: 0,1,2,3,4
    # range(10, 0, -1) # 10,9,8,7,...,1   (paso negativo, cuenta hacia atrás)

# Recorriendo una lista con for 

frutas = ["Manzana","Pera","Uva"]
for fruta in frutas:
    print(fruta)
    #output: manzana,pera,uva



# Recorriendo una string con for 

for letra in "Python":
    print(letra)
    #output: P,y,t,h,o,n

# Con indice Y en el valor 

frutas = ["manzana","pera","uva"]
for indice, fruta in enumerate(frutas):   # enumerate devuelve dos valores: indice y variable ( en este caso,fruta)
    print("indice,fruta")

    













```

## Bucle while 


```python
# No sabemos cuanto se va a repetir, depende de una condicion 

contenedor = 0
while contenedor < 5:
    print(contenedor)
    contenedor += 1    # sin esta linea, contador se quedaria en 0 y el bucle seria infinito
    #output: 0,1,2,3,4

######

clave = input("Ingresa la clave: ")
while clave != "1234":
    clave = input("Ingresa la clave: ")
print("¡Acceso concedido!")

# do while , no existe 

clave = ""
while clave != "1234":
    clave = input("Ingresa la clave: ")
print("¡Acceso concedido!")

```

## Break y continue 


```python
for i in range (10):
    if 1 ==5:
        break
    print(1)
    #output: 0,1,2,3,4

#Continue

for i in range (5):
    if i == 2:
        continue    # salta el valor y continua con el bucle 
    print(i)
    
```

# Colecciones de datos

## Tuplas



```python
# Una tupla es como una lista, pero inmutable: una vez creada, no puedes modificarla (no puedes agregar, quitar ni cambiar elementos).


tupla = (1,2,3)

print(tupla[0]) #Output: 1

# Se usa cuando quieres proteger datos que no deberían modificarse por accidente durante el programa.




```

## Diccionarios


```python
# Un diccionario guarda datos en pares de clave: valor (como una palabra y su definición). En vez de acceder por posición (índice 0, 1, 2...), accedes por nombre (clave).

persona = {"nombre": "Ana","edad":25}

persona["nombre"]
# "Ana"

# .keys() - te da todas las claves
#.values()- te da todos los valores 
#.items()- te da pares (clave,valor)
```

## Sets



```python
# Un set es una colección que elimina automáticamente los duplicados, y además no mantiene ningún orden específico

conjunto = {1,2,3,3,4,2}
print (conjunto)
#output: {1,2,3,4}
```
