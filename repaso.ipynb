# Listas

```python
frutas = ["manzana","pera","uva"]
frutas.append ("kiwi")  #Agrega con .append "kiwi" siempre al final de la lista 

frutas.remove ("pera")  #Elimina "pera" de la lista 

frutas.insert(0,"mango") #Agrega mango a la lista, "0" como indice

frutas.pop()   #Elimina y devuelve el ultimo valor "Kiwi", en este caso, lo elimina y ya no sale en la lista 

frutas.sort()     #sort no devuelve nada, solo modifica la lista (orden acendente en este caso )

frutas

#-----------------------------------------------------------------------------------------------------------#

#len(frutas)   #Cuantos valores hay en la lista, = 3

#"kiwi" in frutas  #False, revisa la lista y devuelve binario si se encuentra 



```
