```python
class Personaje:
    def __init__(self,nombre,vida):
       self.nombre = nombre
       self.vida = vida
    def presentarse(self):
        print(f"Soy {self.nombre} y tengo {self.vida}")
    def recibir_daño(self,daño):
        self.vida -= daño
        print(f"{self.nombre} recibió {daño} de daño. Vida restante: {self.vida}")




v = Personaje("V",100)
print(v.nombre)
print(v.vida)
v.presentarse()
```

```python
class Implante:                                # CLASE
    def __init__(self, nombre, precio, stock): # ATRIBUTO
        self.nombre = nombre
        self.precio = precio
        self.stock = stock
    
    def mostrar_info(self):                    # METODO siempre con self
        print(f"Implante: {self.nombre}")
        print(f"Precio: {self.precio}")
        print(f"Stock: {self.stock}")
    
    def vender(self, cantidad):
        self.stock -= cantidad
        print(f"Vendidos {cantidad} unidades de {self.nombre}. Stock restante: {self.stock}")
    
    def reabastecer(self, cantidad):
        self.stock += cantidad
        print(f"Reabastecido. Nuevo stock de {self.nombre}: {self.stock}")



kiroshi = Implante("Kiroshi Optics", 1500, 10)       # OBJETO
kiroshi.mostrar_info()
kiroshi.vender(3)
kiroshi.reabastecer(5)
       
```
