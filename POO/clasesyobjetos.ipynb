### Que es una clase y objeto?
#### Analogía sencilla: 
#### Una clase es como un plano o molde
#### Un objeto es la casa construida a partir de ese plano

```python
class Galleta:
    pass

una_galleta = Galleta()
otra_galleta = Galleta()
```

##### Clase    = molde                    = class Galleta
##### Objeto   = Algo creado del molde    = otra_galleta= Galleta
##### Atributo = Caracteristica           = self.nombre
##### Metodo   = Accion/funcion           = rodar()

##### Concepto clave: La clase se define solo una vez
##### De una misma clase puedes crear tantos objetos como quieras, cada uno es independiente a la memoria 

#### ATRIBUTOS 
##### Hay dos tipos de atributos 
##### 1.-Atributos de instancia 
##### Son datos que cada objeto tiene por su cuenta, y normalmente se definen dentro de un método especial llamado __init__ (el "constructor").



##### 2.- Atributos de clase 
##### Son datos que comparten todos los objetos de esa clase, porque se definen directamente en el cuerpo de la clase, no dentro de __init__


#### METODOS 
##### Un metodo es simplemente una funcion definida dentro de una clase, que le da un comportamiento al objeto

##### 1.- Metodos de instancia 


```python
class CiberImplante:
    fabricante = "ARASAKA"
def __init__(self,nombre,precio):
       self.nombre = nombre
       self.precio = precio

# Metodo de instancia
def mostrar_info(self):
      print (f"self.nombre - ${self.precio} - Fabricante: {self.fabricante}")

mantis = Ciberimplante("Cuchillas Mantis", 5000)
mantis.mostrar_info()   # Cuchillas Mantis - $5000 - Fabricante: Arasaka
```

##### 1.- Metodos de clase
####  Un método de clase es un método que trabaja con la clase misma, no con un objeto en particular. Se reconoce por dos cosas

#### Lleva el decorador @classmethod arriba.
#### En vez de self, recibe cls (representa a la clase, no al objeto).

```python
class CiberImplante:
    fabricante = "Arasaka"


    def __init__(self,nombre,precio):
        self.nombre = nombre
        self.precio = precio


    @classmethod
    def cambiar_fabricante(cls,nuevo_fabricante):
        cls.fabricante = nuevo_fabricante  #modificando el atributo directamente en la clase


CiberImplante.cambiar_fabricante("Militech")

print(mantis.fabricante)   # Militech
print(optico.fabricante)   # Militech
```
