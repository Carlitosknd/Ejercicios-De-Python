# Ejercicio: Sistema de registro de estudiantes 


```python
print(f"Sistema de registro de estudiantes")
nombre = input(f"Nombre del estudiante: ")
list_calif= []
nota1 = int (input("Primera calificacion:"))
list_calif.append(nota1)
nota2 =  int (input("Segunda calificacion:"))
list_calif.append(nota2)
nota3 =  int (input("Tercera calificacion:"))
list_calif.append(nota3)
#------------------------------------------------#
sum_total = nota1 + nota2 + nota3
prom= sum_total / 3
print (f"Tu promedio es {prom}")

#------------------------------------------------#
status= None

if prom >=90:
    status= True
    print(f"Exelente")
elif prom >=70:
    status= True
    print(f"Aprobado")
else:
    status= False
    print(f"Reprobado")

#------------------------------------------------#

estudiante= {
    "nombre": nombre,
    "Calificaciones": list_calif,
    "Promedio": prom,
    "Estado": status
}
print(estudiante)





    













```
