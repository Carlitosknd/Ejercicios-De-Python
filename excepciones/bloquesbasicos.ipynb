### Bloques basicos 
#### try --> "intenta ejecutar este código, algo puede fallar"
#### except --> "si falla ESTO específico, haz esto otro" 
#### raise --> "yo, manualmente, decido lanzar un error"


```python
def hackear_sistema(nivel_seguridad):
    try:   # algo puede fallar
        resultado = 100 / nivel_seguridad
        print(f"Sistema hackeado. Puntos de daño: {resultado}") #ok
    except ZeroDivisionError: #tipo de error y que imprimir 
        print("ICE detectado: nivel de seguridad no puede ser 0")

# 

```

#### else y finally

#### else: código que solo se ejecuta si el try NO tuvo errores.
#### finally: código que se ejecuta SIEMPRE — con o sin error. Se usa muchísimo para cerrar archivos, conexiones a bases de datos, liberar recursos

```python
def conectar_netrunner(cyberdeck_activo):
    try:
        if not cyberdeck_activo: #si no esta activo
            raise ConnectionError("Cyberdeck desconectado") #error de conexion 
        print("Conectando a la red...")
    except ConnectionError as e: #si falla,toma el error y guardalo como "e"
        print(f"Fallo la conexion: {e}") 
    else:
        print("Conexion exitosa, sin errores")
    finally:
        print("Cerrando el socket de red") # SIEMPRE corre, haya error o no

# except ConnectionError as e no revisa qué línea falló ni dónde — solo dice "si llega volando un error de tipo ConnectionError desde cualquier parte del try, yo lo atrapo y lo guardo en e". 
# No le importa si vino de la línea 2, 3 o 5 del try.
```

#### Excepciones mas usadas
##### ValueError, TypeError, KeyError, IndexError, ZeroDivisionError, FileNotFoundError, AttributeError.

##### Nunca usar except: a secas (sin especificar el tipo). Atrapa TODO, incluso errores de programación, como un KeyboardInterrupt o un bug real.

  

```python
class SaldoInsuficienteError(Exception): #clase 
    pass

def retirar_eddies(saldo, cantidad): #funcion de dos parametros 
    try:     # intenta, algo puede fallar
        if cantidad < 0:
            raise ValueError("No puede ser saldo negativo") # manualmente lanzo el error como Value  
        if cantidad > saldo:
            raise SaldoInsuficienteError("Cantidad mayor al saldo disponible") # ///
    except ValueError as e: # si falla esto especifico, guardalo e imprimelo
        print(f"Error: {e}")
    except SaldoInsuficienteError as e: # ///
        print(f"Error: {e}")
    else:   
        return saldo - cantidad
    finally:
        print("Cierre total")


print(retirar_eddies(100, 30))   # caso normal

```
