```python
class StockInsuficienteError(Exception): #clase 
    pass


def prestar_libro (copias_disponibles, cantidad_solicitada):

    try:
        if cantidad_solicitada <=0:
            raise ValueError("Cantidad negativa o igual a cero")

        if cantidad_solicitada > copias_disponibles:
            raise StockInsuficienteError ("Cantidad solicitada es mayor que copias disponibles")

    except ValueError as e: # ///
        print(f"Error: {e}")

    except StockInsuficienteError as e: # ///
        print(f"Error: {e}")
    
    else:  # "si el try no tuvo ningún error, entonces...".
        return(copias_disponibles - cantidad_solicitada)
    finally:
        print("Transacción de préstamo finalizada")



print(prestar_libro(10, 3))    # caso normal
print(prestar_libro(10, 0))    # cantidad en cero
print(prestar_libro(10, 15))   # más de lo disponible
    
        
        
   
```
