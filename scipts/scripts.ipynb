# Scripts
## Un script es simplemente un archivo .py que contiene código que se ejecuta de principio a fin, como un programa.
Tipicamente son : python nombre_archivo.py

import sys 
sys es un módulo (una librería que ya viene incluida en Python) que te da acceso a información relacionada con el sistema

sys.argv es una lista permite analizar los datos que se envian a un script durante la ejecucion atraves de la linea de comandoos 

### En el ejemplo de sumar.py 

Paso 1 

python sumar.py 5 3

Paso 2 — Python separa esto por espacios y lo guarda en una lista, automáticamente:

python
sys.argv = ["sumar.py", "5", "3"]

###             [0]        [1]  [2]

Paso 3 — Extraes cada dato por su índice

python
sys.argv[0]   # "sumar.py"  <- el nombre del script, casi nunca lo usas

sys.argv[1]   # "5"          <- primer argumento 

sys.argv[2]   # "3"          <- segundo argumento
