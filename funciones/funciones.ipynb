Return


Regla mental: si necesitas usar el resultado de la función más adelante (guardarlo, sumarlo, pasarlo a otra función), necesitas return. Si solo quieres mostrar algo al usuario y ya, print basta.


```python
def sumar (a,b): #dos parametros
    return a + b  

total= sumar(5,10) #ingresamos parametros a,b
total_con_impuesto = total * 1.15  #(a,b*1.15)
print(total_con_impuesto)

```

### Parametros con valor por defecto
#### Una regla importante que a veces se pasa por alto: los parámetros con valor por defecto siempre van al final, después de los que no lo tienen.Y se pueden sobreescribir


```python
def presentar (nombre,ciudad="Monterrey"):
    print(f"{nombre} vive en {ciudad}")

presentar("Ana")  
presentar("Luis", "CDMX")
```

### Multiples return 
#### Una función puede tener más de un return, pero en cuanto Python ejecuta uno, la función termina ahí mismo — no sigue leyendo el código de abajo.
#### Cualquier línea después de un return que se ejecute, jamás corre.

```python
def clasificar_edad(edad):
    if edad < 13:
        return "niño"
    elif edad < 18:
        return "adolecente"
    else:
        return"adulto"

print(clasificar_edad(17))



    
```

### Scope
#### Las variables creadas dentro de una función solo existen ahí adentro. Fuera de la función, no existen.

```python
def calcular():
    resultado = 100  #variable local
    return resultado

calcular()

#print(resultado) , no esta definido 
```

#### En cambio, una variable creada fuera de las funciones (variable global) sí se puede leer desde dentro:

```python
mensaje = "Hola desde afuera"

def saludar():
    print(mensaje)

saludar()

```

### Funciones que llaman a otras funciones 
#### Una función puede usar el resultado de otra función adentro de su propio código
#### En vez de repetir la lógica de "elevar al cuadrado" dos veces, se escribe una vez en cuadrado() y la reutilizas.

```python
def cuadrado(numero):
    return numero ** 2

def suma_de_cuadrados(a,b):
    return cuadrado(a) + cuadrado (b)

print (suma_de_cuadrados(3,4)) # cuadrado(3)=9, cuadrado(4)=16 -> 25
```

### Funciones recursivas 

#### Funciones que se vuelven a llaman y acaban


```python
def cuenta_atras(num):
    num -= 1
    if num > 0:
        print (num)
        cuenta_atras(num)
    else:
        print("Booom")

cuenta_atras(5)
```
