```python
def costo_reparacion(nivel_daño): # Recibe un numero
    return (nivel_daño * 15)


def diagnostico_implante(nivel_daño):
    if costo_reparacion(nivel_daño) > 1000:
        return ("CRITICO, REMPLAZO TOTAL REQUERIDO")
    elif 300 <= costo_reparacion(nivel_daño) <= 1000: #rango
          return ("DAÑO MODERADO")
    else:
         return ("DAÑO LEVE")


print(diagnostico_implante(200))





# costo_reparacion           # solo el nombre, no hace nada
#(costo_reparacion)         # sigue siendo solo el nombre, los paréntesis aquí no hacen nada
#costo_reparacion(nivel_daño)   # ESTO sí la ejecuta con nivel_daño como argumento


```

```python
def calcular_recompensa (dificultad, bono = 0): 
    return (dificultad * 200 + bono) # return (dificultad * 200, bono) crea una tupla con ,


def evaluar_mision(dificultad,bono = 0):
    if calcular_recompensa(dificultad,bono) > 1500:
        return("MISION LEGENDARIA - PAGO EXCELENTE")
    elif 500 <= calcular_recompensa(dificultad,bono) <= 1500:
        return ("MISION ESTANDAR - PAGO ACEPTABLE")
    else:
        return ("MISION BAJA - PAGO BAJO")


print(evaluar_mision(10, 200))   
print(evaluar_mision(4))        
print(evaluar_mision(1))         


```

```python
def valor_pieza(tipo,cantidad = 1):
    if tipo == "comun":
        return(100 * cantidad)
    elif tipo =="raro":
        return(500 * cantidad)
    else:
        return(2000 * cantidad)

    # tenemos que regresar el numero multiplicado por cantidad , (dos return no funcionaba)
    

def negociar_venta(tipo,cantidad = 1):  
    if valor_pieza(tipo,cantidad) > 3000:
        return ("TRATO PREMIUM - FIXER PAGA EN EFECTIVO")
    elif 500 <= valor_pieza(tipo,cantidad) <= 3000 :
        return("TRATO NORMAL -  PAGO ESTANDAR")
    else:
        return("TRATO MENOR - NO VALE")
    

print(negociar_venta("raro",2))
```
