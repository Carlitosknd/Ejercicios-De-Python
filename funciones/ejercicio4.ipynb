#### Ejercicio: Sistema de hackeo de Night City

#### Crea dos funciones:

#### 1. calcular_daño_hackeo(nivel_hacker, dificultad=5)

#### Recibe el nivel_hacker (un número) y la dificultad del sistema a hackear (valor por defecto de 5).
#### Debe regresar (con return) el daño causado al sistema.
#### Fórmula: nivel_hacker * 10 - dificultad

#### 2. intentar_hackeo(nivel_hacker, dificultad=5)

#### Debe llamar a calcular_daño_hackeo para obtener el daño.
#### Si el daño es mayor a 50, debe regresar el texto "ACCESO CONCEDIDO - ICE eliminado"
#### Si el daño es 50 o menos, debe regresar el texto "ACCESO DENEGADO - Trace iniciado"
#### Debe tener dos return distintos (uno para cada caso).
#### Reglas
#### Solo programa las dos funciones, sin código adicional.
#### No uses print dentro de ninguna, solo return.

```python
def calcular_damage_hackeo(nivel_hacker, dificultad=5):
    return nivel_hacker * 10 - dificultad


def intentar_hackeo(nivel_hacker, dificultad=5):
    if calcular_damage_hackeo(nivel_hacker, dificultad) > 50:
        return "ACCESO CONCEDIDO - ICE ELIMINADO"
    else:
        return "ACCESO DENEGADO - TRACE INICIADO"


print(intentar_hackeo(8))          # ACCESO CONCEDIDO
print(intentar_hackeo(3))          # ACCESO DENEGADO
print(intentar_hackeo(6, 15))      # ACCESO DENEGADO
print(intentar_hackeo(10, 20))     # ACCESO CONCEDIDO
```
